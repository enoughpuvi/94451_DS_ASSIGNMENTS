# Que 1
# 1. NewspaperData.csv

## Import libraries and read dataset

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import warnings
warnings.filterwarnings(action="ignore")
```

```python
df= pd.read_csv('NewspaperData.csv')
df.head()
```

## EDA

```python
df.info()
```

```python
df.isna().sum() ## null values
```

## Split data in x,y

```python
x=df.drop(['sunday','Newspaper'],axis=1)
y=df['sunday']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, train_size=0.8, random_state=123)
```

## Model building and training

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x,y)
```

```python
pred = model.predict([[500]])
print(f"If newspaper has a price of 500 daily,on sunday it will have price = {pred[0]:.2f}")
```

## Visualization

```python
plt.scatter(df['daily'],df['sunday'],label='sunday')
plt.plot(df['daily'],model.predict(x),color="red", label ="best fit regression line")
plt.scatter(df['daily'],model.predict(x),color="red")
plt.xlabel('Daily')
plt.ylabel("Sunday")
plt.title('Daily vs sunday')
plt.legend()
```

## Accuracy

```python
model.score(x,y)
```

# Predict the data

```python
y_prediction = model.predict(x_test)
print(x_test)
```

```python
y_test
```

```python
y_prediction
```

## Performance check

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
mae = mean_absolute_error(y_test,y_prediction)
print(f"Mean Absoulte Error = {mae}")
```

```python
mse=mean_squared_error(y_test,y_prediction)
print(f"Mean Squared Error = {mse}")
```

```python
rmse=np.sqrt(mse)
print(f"Root Mean Square = {rmse}")
```

```python
r2=r2_score(y_test,y_prediction)
print(f"R2 score = {r2}")
```
