# Data cleansing

```python
df['population_density'] = df['population_density'].astype(str)
df['population_density'] = df['population_density'].str.replace(r'[^0-9]', '', regex=True)
df['population_density'] = pd.to_numeric(df['population_density'], errors='coerce')
```

```python
df['Gross_weight'] = df['Gross_weight'].astype(str)
df['Gross_weight'] = df['Gross_weight'].str.replace(r'[^0-9.]', '', regex=True)
df['Gross_weight'] = pd.to_numeric(df['Gross_weight'], errors='coerce')
```

```python
df['Length'] = df['Length'].astype(str)
df['Length'] = df['Length'].str.replace(r'[^0-9.]', '', regex=True)
df['Length'] = pd.to_numeric(df['Length'], errors='coerce')
```

```python
df['width'] = df['width'].astype(str)
df['width'] = df['width'].str.replace(r'[^0-9.]', '', regex=True)
df['width'] = pd.to_numeric(df['width'], errors='coerce')
```

# Drop unnecessary columns

```python
df=df.drop(['policy_id'],axis=1)
```

# Split Columns

```python
df[['torque', 'rpm_torque']] = df['max_torque'].str.extract(r'(\d+)Nm@(\d+)rpm')
df['torque'] = df['torque'].astype(float)
df['rpm_torque'] = df['rpm_torque'].astype(int)
```

```python
df[['power', 'rpm_power']] = df['max_power'].str.extract(r'([\d.]+)bhp@(\d+)rpm')
df['power'] = df['power'].astype(float)
df['rpm_power'] = df['rpm_power'].astype(int)
```

```python
df.drop(['max_torque','max_power'],axis=1,inplace=True)
```

```python
df.columns
```

# Add columns

```python
df['power_to_rpm']=df['power']/df['rpm_power']
```

```python
df['torque_to_rpm']=df['torque']/df['rpm_torque']
```

# Encoding

## One hot encoding

```python
from sklearn.preprocessing import OneHotEncoder
categorical_columns=['area_cluster', 'segment','fuel_type','engine_type','rear_brakes_type','transmission_type','steering_type','model']
print(f"Original Employee Data:\n{df}\n")
df_pandas_encoded = pd.get_dummies(df, columns=['area_cluster','model', 'segment','fuel_type','engine_type','rear_brakes_type','transmission_type','steering_type'], drop_first=True)
print(f"One-Hot Encoded Data using Pandas:\n{df_pandas_encoded}\n")
encoder = OneHotEncoder(sparse_output=False)
one_hot_encoded = encoder.fit_transform(df[categorical_columns])
one_hot_df = pd.DataFrame(one_hot_encoded,
                          columns=encoder.get_feature_names_out(categorical_columns))
df = pd.concat([df.drop(categorical_columns, axis=1), one_hot_df], axis=1)
```

# Model Building

```python
x=df.drop('is_claim',axis=1)
y=df['is_claim']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test=train_test_split(x,y,train_size=0.7,random_state=123456,stratify=y)
```

# Resampling

```python
print("Before SMOTE:")
print(y_train.value_counts())
```

```python
from imblearn.over_sampling import SMOTE
smote = SMOTE(random_state=42)
x_train, y_train = smote.fit_resample(
    x_train, y_train
)
```

```python
print("\nAfter SMOTE:")
print(y_train.value_counts())
```

# Hyperparameter

```python
from sklearn.model_selection import GridSearchCV
```

## Logistic Regression model

```python
from sklearn.linear_model import LogisticRegressionCV

def create_logistic_model():
    model = LogisticRegressionCV(
        Cs=[0.01, 0.1, 1, 10],
        penalty='l2',
        solver='liblinear',
        scoring='f1',
        cv=3,
        max_iter=1000,
        n_jobs=-1,
        random_state=42
    )

    model.fit(x_train, y_train)
    return model
```

## KNN Model

```python
from sklearn.neighbors import KNeighborsClassifier
def create_knn_model():
    knn = KNeighborsClassifier()
    param_grid={
        'n_neighbors':[3,5,7,9]
    }
    model = GridSearchCV(knn,param_grid,cv=5)
    model.fit(x_train,y_train)
    return model
```

## SVM Model 

```python
from sklearn.svm import SVC
def create_svm_model():
    # model=SVC()
    # model.fit(x_train,y_train)
    # return model
    param_grid = {
        'C': [0.1, 1, 10],
        'kernel': ['rbf'],
        'gamma': ['scale', 'auto']
    }

    grid = GridSearchCV(
        estimator=SVC(),
        param_grid=param_grid,
        cv=5,
        scoring='f1',
        n_jobs=-1
    )

    grid.fit(x_train, y_train)

    # Best SVM model
    model = grid.best_estimator_

    return model
```

## Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
def create_decision_tree_model():
    # model=DecisionTreeClassifier(max_depth=50)
    # model.fit(x_train,y_train)
    # return model
    param_grid = {
        'criterion': ['gini', 'entropy'],
        'max_depth': [ 10, 20, 30, 50],
        'min_samples_split': [2, 5, 10],
        'min_samples_leaf': [1, 2, 4]
    }

    grid = GridSearchCV(
        estimator=DecisionTreeClassifier(random_state=42),
        param_grid=param_grid,
        cv=5,
        scoring='f1',
        n_jobs=-1
    )

    grid.fit(x_train, y_train)

    # Best Decision Tree model
    model = grid.best_estimator_

    return model
```

# Model Evaluation

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score,confusion_matrix
def evaluate_model_on_test_data(model):
    y_pred = model.predict(x_test)
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average="weighted")
    recall = recall_score(y_test, y_pred, average="weighted")
    f1 = f1_score(y_test, y_pred, average="weighted")
    cm = confusion_matrix(y_test, y_pred)
    print(f"Confusion Matrix: ")
    print(cm)
    return f"{accuracy:.2f}", f"{precision:.2f}", f"{recall:.2f}", f"{f1:.2f}"
```

```python
# create the models
models = []
```

```python
models.append({"name": "SVM", "model": create_svm_model()})  
```

# Summary

```python
# Evaluate models
for info in models:
    accuracy,precision,recall,f1=evaluate_model_on_test_data(info["model"])
    info['accuracy']=accuracy
    info["precision"] = precision
    info["recall"] = recall  
    info["f1"] = f1
    
# report
df_evaluation_report_test = pd.DataFrame(models, columns=["name", "accuracy", "precision", "recall", "f1"])
df_evaluation_report_test
```
